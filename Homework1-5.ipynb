Решить классически и смоделировать решение задачи.

> В одной урне 15 белых и 5 черных шаров, а в другой - 6 белых и 1 черный шар. Из первой урны случайным образом вынимают 3 шара и опускают во вторую урну. После этого из второй урны также случайно вынимают 1 шар. Найти вероятность того, что шар, вынутый из второй урны - белый.

Рассмотрим событие $A$ — из второй урны извлечен белый шар.

```python
#Классическое решение задачи
p_A = (sc.comb(15, 0)*sc.comb(5, 3)/sc.comb(20, 3)*sc.comb(6, 1)/sc.comb(10, 1)+
sc.comb(15, 1)*sc.comb(5, 2)/sc.comb(20, 3)*sc.comb(7, 1)/sc.comb(10, 1)+
sc.comb(15, 2)*sc.comb(5, 1)/sc.comb(20, 3)*sc.comb(8, 1)/sc.comb(10, 1)+
sc.comb(15, 3)*sc.comb(5, 0)/sc.comb(20, 3)*sc.comb(9, 1)/sc.comb(10, 1))
print(f"{p_A:.4f}")
```

```python
import random

# Количество итераций для моделирования
n_trials = 100000
successes = 0

# Исходный состав урн: 'W' - белый шар, 'B' - черный шар
urn1 = ['W'] * 15 + ['B'] * 5
urn2_base = ['W'] * 6 + ['B'] * 1

for _ in range(n_trials):
    # Выбираем 3 шара из первой урны без возвращения
    drawn = random.sample(urn1, 3)
    
    # Добавляем выбранные шары во вторую урну
    urn2 = urn2_base + drawn
    
    # Выбираем 1 случайный шар из второй урны
    final_ball = random.choice(urn2)
    
    # Проверяем условие извлечения белого шара
    if final_ball == 'W':
        successes += 1

# Вычисляем смоделированную вероятность
simulated_prob = successes / n_trials
round(simulated_prob, 4)
```

### Задача 2.9
> В среднем 20% пакетов акций на аукционах продаются по первоначально заявленной цене. Найти вероятность того,
что из 9 пакетов акций в результате торгов по первоначально заявленной цене: 1) не будут проданы 5 пакетов; 2) будет продано:
а) менее 2 пакетов; б) не более 2 пакетов; в) хотя бы 2 пакета; г) наивероятнейшее число пакетов.

* Рассмотрим событие A - по первоначально заявленной цене не будут проданы 5 пакетов
* Рассмотрим событие Bа - по первоначально заявленной цене будет продано менее 2 пакетов
* Рассмотрим событие Bб - по первоначально заявленной цене будет продано не более 2 пакетов
* Рассмотрим событие Bв - по первоначально заявленной цене будет продано хотя бы 2 пакета
* Рассмотрим событие Bг - по первоначально заявленной цене будет продано наивероятнейшее число пакетов

```python
from scipy.stats import binom

n, p = 9, 0.2
# P(A): k = 4
p_1 = binom.pmf(4, n, p)
# P(Bа): k < 2
p_2a = binom.cdf(1, n, p)
# P(Bб): k <= 2
p_2b = binom.cdf(2, n, p)
# P(Bв): k >= 2
p_2v = 1 - binom.cdf(1, n, p)
# P(Bг): наивероятнейшее число
m0_min = n * p - (1 - p)
m0_max = n * p + p

print(f"1) {round(p_1, 4)}")
print(f"2а) {round(p_2a, 4)}")
print(f"2б) {round(p_2b, 4)}")
print(f"2в) {round(p_2v, 4)}")
print(f"2г) [{m0_min}, {m0_max}]")
```

### Задача 2.10
> Завод отправил на базу 10_000 стандартных изделий. Среднее число изделий, повреждаемых при транспортировке,
составляет 0,02%. Найти вероятность того, что из 10_000 изделий: 1) будет повреждено: а) 3; б) по крайней мере 3; 2) не будет повреждено: а) 9997; б) хотя бы 9997. 

* Рассмотрим событие Aа - будет повреждено 3 изделия
* Рассмотрим событие Aб - будет повреждено по крайней мере 3 изделия
* Рассмотрим событие Bа - не будет повреждено 9997 изделий
* Рассмотрим событие Bб - не будет повреждено хотя бы 9997 изделий

```python
from scipy.stats import poisson

mu = 10000 * 0.0002 # lambda = 2.0

# P(Aа): k = 3
p_1a = poisson.pmf(3, mu)
# P(Aб): k >= 3
p_1b = 1 - poisson.cdf(2, mu)
# P(Bа): k = 3 (аналогично 1а)
p_2a = poisson.pmf(3, mu)
# P(Bб): k <= 3
p_2b = poisson.cdf(3, mu)

print(f"1а) {round(p_1a, 4)}")
print(f"1б) {round(p_1b, 4)}")
print(f"2а) {round(p_2a, 4)}")
print(f"2б) {round(p_2b, 4)}")
```

### Задача 2.11
> По результатам проверок налоговыми инспекциями установлено, что в среднем каждое второе малое предприятие региона имеет нарушение финансовой дисциплины. Найти вероятность того, что из 1000 зарегистрированных в регионе малых предприятий имеют нарушения финансовой дисциплины: а) 480 предприятий; б) наивероятнейшее число предприятий; в) не менее 480; г) от 480 до 520. 

* Рассмотрим событие A - 480 малых предприятий имеют нарушения финансовой дисциплины
* Рассмотрим событие B - наивероятнейшее число малых предприятий имеют нарушения финансовой дисциплины
* Рассмотрим событие C - не менее 480 малых предприятий имеют нарушения финансовой дисциплины
* Рассмотрим событие D - от 480 до 520 малых предприятий имеют нарушения финансовой дисциплины

```python
import math
from scipy.stats import norm

n, p = 1000, 0.5
np_val = n * p
std = math.sqrt(n * p * (1 - p))

# а) k = 480 (локальная теорема)
x_480 = (480 - np_val) / std
p_a = norm.pdf(x_480) / std

# б) наивероятнейшее число
m0 = int(np_val)

# в) от 480 до 1000 (интегральная теорема)
x_1000 = (1000 - np_val) / std
p_v = norm.cdf(x_1000) - norm.cdf(x_480)

# г) от 480 до 520 (интегральная теорема)
x_520 = (520 - np_val) / std
p_g = norm.cdf(x_520) - norm.cdf(x_480)

print(f"а) {round(p_a, 4)}")
print(f"б) m0 = {m0}")
print(f"в) {round(p_v, 4)}")
print(f"г) {round(p_g, 4)}")
```
